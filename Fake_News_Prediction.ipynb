```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC

from sklearn.metrics import accuracy_score, confusion_matrix

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
fake = pd.read_csv('/content/Fake.csv', on_bad_lines='skip', engine='python')
true = pd.read_csv('/content/True.csv', on_bad_lines='skip', engine='python')

# Add labels
fake['label'] = 1
true['label'] = 0

# Combine datasets
df = pd.concat([fake, true], axis=0)

# Shuffle dataset
df = df.sample(frac=1, random_state=2).reset_index(drop=True)

print("Dataset Shape:", df.shape)
```

```python
df.to_csv('train.csv', index=False)
```

```python
df['content'] = df['title'] + " " + df['text']

X = df['content']
Y = df['label']
```

```python
df['content'] = df['title'] + " " + df['text']

X = df['content']
Y = df['label']
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=2)
```

```python
vectorizer = TfidfVectorizer(max_features=5000)
X_train_vectorized = vectorizer.fit_transform(X_train)
X_test_vectorized = vectorizer.transform(X_test)

lr_model = LogisticRegression(max_iter=1000)
lr_model.fit(X_train_vectorized, Y_train)

lr_train_pred = lr_model.predict(X_train_vectorized)
lr_test_pred = lr_model.predict(X_test_vectorized)

lr_train_acc = accuracy_score(Y_train, lr_train_pred)
lr_test_acc = accuracy_score(Y_test, lr_test_pred)

print("\nLogistic Regression:")
print("Training Accuracy:", lr_train_acc)
print("Testing Accuracy:", lr_test_acc)
```

```python
svm_model = SVC(kernel='linear')
svm_model.fit(X_train_vectorized, Y_train)

svm_train_pred = svm_model.predict(X_train_vectorized)
svm_test_pred = svm_model.predict(X_test_vectorized)

svm_train_acc = accuracy_score(Y_train, svm_train_pred)
svm_test_acc = accuracy_score(Y_test, svm_test_pred)

print("\nSVM Model:")
print("Training Accuracy:", svm_train_acc)
print("Testing Accuracy:", svm_test_acc)
```

```python
print("\nModel Comparison:")
print("--------------------------")
print("Logistic Regression Test Accuracy:", lr_test_acc)
print("SVM Test Accuracy:", svm_test_acc)
```

```python
def predict_news(text):
    input_data = vectorizer.transform([text])
    prediction = svm_model.predict(input_data)

    if prediction[0] == 1:
        return "Fake News"
    else:
        return "Real News"
```

```python
sample = "Breaking news about government policy changes"
print("\nPrediction:", predict_news(sample))
```
